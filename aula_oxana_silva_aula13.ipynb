```python

# -------- Etapa 2 - Inspecao e Tratamento dos Dados --------

import pandas as pd

# Carregando a base de dados
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
dados = pd.read_csv(url)

# 1. Inspecionando os primeiros registros
print("Primeiros registros da base:")
print(dados.head())

# 2. Verificando informacoes da base
print("\nInformacoes da base:")
dados.info()

# 3. Verificando valores ausentes
print("\nValores ausentes por coluna:")
print(dados.isnull().sum())

# 4. Removendo registros duplicados
dados = dados.drop_duplicates()

# 5. Removendo valores ausentes
dados = dados.dropna()

# 6. Verificando dados inconsistentes
# Removendo medidas iguais ou menores que zero
colunas = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]

for coluna in colunas:
    dados = dados[dados[coluna] > 0]

# 7. Mostrando o resultado final
print("\nBase de dados apos o tratamento:")
print(dados.head())

print("\nValores ausentes apos o tratamento:")
print(dados.isnull().sum())

print("\nQuantidade final de linhas e colunas:")
print(dados.shape)

print("\nTratamento de dados concluido!")
```

```python

# -------- Etapa 3 - Filtragem e Exportacao dos Dados --------

import pandas as pd

# Carregando a base de dados
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
dados = pd.read_csv(url)

# Removendo valores ausentes e duplicados
dados = dados.dropna()
dados = dados.drop_duplicates()

# Criando um filtro para selecionar apenas flores Setosa
filtro = dados[dados["species"] == "setosa"]

# Mostrando os dados filtrados
print("Flores da especie Setosa:")
print(filtro.head())

# Mostrando a quantidade de registros
print("\nQuantidade de flores encontradas:")
print(len(filtro))

# Salvando os dados em um novo arquivo CSV
filtro.to_csv("iris_setosa.csv", index=False)

print("\nArquivo iris_setosa.csv salvo com sucesso!")
```

```python

# -------- Etapa 2 - Criacao de Graficos --------

# Importando as bibliotecas
import pandas as pd
import matplotlib.pyplot as plt

# Carregando a base de dados limpa
dados = pd.read_csv("iris_setosa.csv")

# Calculando a media das medidas das petalas
comprimento = dados["petal_length"].mean()
largura = dados["petal_width"].mean()

# Criando os dados para o grafico
categorias = ["Comprimento", "Largura"]
medias = [comprimento, largura]

# Criando o grafico de barras
plt.bar(categorias, medias)

# Adicionando titulo e nomes dos eixos
plt.title("Media das Medidas das Petalas - Setosa")
plt.xlabel("Medidas")
plt.ylabel("Media em centimetros")

# Exibindo os valores nas barras
for i, valor in enumerate(medias):
    plt.text(i, valor, round(valor, 2), ha="center")

# Exibindo o grafico
plt.show()
```

```python

# -------- Etapa 3 - Profissionalizacao do Grafico --------

import pandas as pd
import matplotlib.pyplot as plt

# Carregando a base de dados
dados = pd.read_csv("iris_setosa.csv")

# Calculando as medias
comprimento = dados["petal_length"].mean()
largura = dados["petal_width"].mean()

# Organizando os dados
categorias = ["Comprimento", "Largura"]
medias = [comprimento, largura]

# Definindo as cores
cores = ["royalblue", "orange"]

# Criando o grafico
plt.figure(figsize=(8, 5))

plt.bar(categorias, medias, color=cores)

# Adicionando titulo
plt.title("Comparacao das Medidas das Petalas - Setosa",
          fontsize=14)

# Identificando os eixos
plt.xlabel("Medidas das Petalas", fontsize=12)
plt.ylabel("Media em centimetros", fontsize=12)

# Exibindo os valores nas barras
for i, valor in enumerate(medias):
    plt.text(i, valor + 0.02, round(valor, 2),
             ha="center", fontsize=11)

# Adicionando grade
plt.grid(axis="y", linestyle="--", alpha=0.4)

# Ajustando a visualizacao
plt.tight_layout()

# Exibindo o grafico
plt.show()
```
