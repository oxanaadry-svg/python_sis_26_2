```python

# Etapa 1: Estrutura em Memoria

# Cria um dicionario com as configuracoes do cluster
cluster = {
    "nome_cluster": "Cluster_Principal",
    "versao_so": "Ubuntu Server 22.04",

    # Lista com 3 servidores
    "servidores": [
        {
            "servidor": "Servidor 1",
            "ip": "192.168.0.10",
            "portas": [80, 443, 22]
        },
        {
            "servidor": "Servidor 2",
            "ip": "192.168.0.11",
            "portas": [80, 443, 3306]
        },
        {
            "servidor": "Servidor 3",
            "ip": "192.168.0.12",
            "portas": [22, 8080, 5432]
        }
    ]
}

# Exibe as configuracoes do cluster
print(cluster)

```

```python

# Etapa 2: Exportacao JSON

# Importa a biblioteca json
import json

# Abre o arquivo JSON para escrita
with open("cluster_config.json", "w", encoding="utf-8") as arquivo:

    # Salva o dicionario de forma organizada
    json.dump(cluster, arquivo, indent=4, ensure_ascii=False)

# Exibe uma mensagem de confirmacao
print("Arquivo cluster_config.json criado com sucesso!")
```

```python

# Etapa 3: Exportacao CSV

# Importa a biblioteca csv
import csv

# Abre o arquivo CSV para escrita
with open("servidores_cluster.csv", "w", newline="", encoding="utf-8") as arquivo:

    # Define as colunas do arquivo
    colunas = ["Servidor", "IP", "Portas"]

    # Cria o escritor CSV com delimitador ;
    escritor = csv.DictWriter(
        arquivo,
        fieldnames=colunas,
        delimiter=";"
    )

    # Grava o cabecalho
    escritor.writeheader()

    # Percorre a lista de servidores
    for servidor in cluster["servidores"]:

        # Organiza os dados de cada servidor
        dados = {
            "Servidor": servidor["servidor"],
            "IP": servidor["ip"],
            "Portas": ", ".join(map(str, servidor["portas"]))
        }

        # Grava os dados no arquivo CSV
        escritor.writerow(dados)

# Exibe uma mensagem de confirmacao
print("Arquivo servidores_cluster.csv criado com sucesso!")
```

```python

# Etapa 2: Criacao da planilha

# Importa a biblioteca Workbook
from openpyxl import Workbook

# Cria uma nova pasta de trabalho
planilha = Workbook()

# Cria uma aba chamada Servidores
aba = planilha.active
aba.title = "Servidores"

# Adiciona o cabecalho da planilha
aba.append([
    "Hostname",
    "IP",
    "Memoria (GB)",
    "Uso de Disco (%)",
    "Status"
])

# Exibe mensagem de confirmacao
print("Planilha criada com sucesso!")
```

```python

# Etapa 3: Insercao de dados

# Cria uma lista com os dados dos servidores
servidores = [
    ["Servidor01", "192.168.0.10", 16, 65, "Ativo"],
    ["Servidor02", "192.168.0.11", 32, 85, "Ativo"],
    ["Servidor03", "192.168.0.12", 8, 45, "Ativo"],
    ["Servidor04", "192.168.0.13", 16, 90, "Ativo"]
]

# Percorre a lista de servidores
for servidor in servidores:

    # Adiciona os dados na planilha
    aba.append(servidor)

# Exibe mensagem de confirmacao
print("Dados dos servidores inseridos com sucesso!")
```

```python

# Etapa 4: Avisos e salvamento da planilha

# Percorre os servidores da planilha
for linha in range(2, aba.max_row + 1):

    # Verifica o uso de disco
    uso_disco = aba.cell(row=linha, column=4).value

    # Verifica se o uso e maior ou igual a 80%
    if uso_disco >= 80:

        # Obtem o nome do servidor
        hostname = aba.cell(row=linha, column=1).value

        # Exibe um aviso no terminal
        print("ATENCAO:", hostname,
              "precisa de expansao de armazenamento!")

# Salva a planilha Excel
planilha.save("inventario_servidores.xlsx")

# Exibe mensagem de confirmacao
print("\nPlanilha inventario_servidores.xlsx salva com sucesso!")
```

```python

# Etapa 1: Leitura do arquivo log.txt

# Tenta abrir o arquivo para leitura
try:

    # Abre o arquivo utilizando o bloco with
    with open("log.txt", "r", encoding="utf-8") as arquivo:

        # Le o conteudo do arquivo
        conteudo = arquivo.read()

        # Exibe o conteudo no terminal
        print("Conteudo do arquivo log.txt:")
        print(conteudo)

except FileNotFoundError:

    # Exibe mensagem caso o arquivo nao exista
    print("Erro: O arquivo log.txt nao foi encontrado.")
```

```python

# Etapa 2: Extracao do nivel do evento

try:

    # Abre o arquivo para leitura
    with open("log.txt", "r", encoding="utf-8") as arquivo:

        # Percorre cada linha do arquivo
        for linha in arquivo:

            # Verifica o nivel do evento
            if "INFO" in linha:
                nivel = "INFO"

            elif "AVISO" in linha:
                nivel = "AVISO"

            elif "ERRO" in linha:
                nivel = "ERRO"

            else:
                nivel = "Nao identificado"

            # Exibe o nivel encontrado
            print("Nivel do evento:", nivel)

except FileNotFoundError:
    print("Erro: O arquivo log.txt nao foi encontrado.")
```

```python

# Importa as bibliotecas
from pathlib import Path
from openpyxl import Workbook

# Define o arquivo de log
arquivo_log = Path("log.txt")

# Verifica se o arquivo existe
if not arquivo_log.exists():
    print("Erro: O arquivo log.txt nao foi encontrado.")

else:
    # Cria uma nova planilha
    planilha = Workbook()
    aba = planilha.active
    aba.title = "Auditoria de Logs"

    # Cria os titulos das colunas
    aba.append(["Data/Hora", "Tipo de Evento", "Mensagem"])

    # Abre o arquivo de log para leitura
    with open(arquivo_log, "r", encoding="utf-8") as arquivo:

        # Percorre cada linha do arquivo
        for linha in arquivo:

            linha = linha.strip()

            # Ignora linhas vazias
            if linha == "":
                continue

            # Separa as informacoes da linha
            partes = linha.split(" ", 2)

            # Verifica se a linha possui as informacoes
            if len(partes) >= 3:

                data_hora = partes[0] + " " + partes[1]
                restante = partes[2]

                # Identifica o tipo de evento
                if "INFO" in restante:
                    tipo = "INFO"
                elif "AVISO" in restante:
                    tipo = "AVISO"
                elif "ERRO" in restante:
                    tipo = "ERRO"
                else:
                    tipo = "DESCONHECIDO"

                # Extrai a mensagem
                mensagem = restante.replace(tipo, "", 1).strip()

                # Adiciona os dados na planilha
                aba.append([data_hora, tipo, mensagem])

    # Salva a planilha no diretorio de execucao
    planilha.save("auditoria_logs.xlsx")

    print("Planilha auditoria_logs.xlsx criada com sucesso!")

```
