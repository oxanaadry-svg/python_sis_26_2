```python
# Etapa 1: Inicializacao do Log (Modo Escrita - w)

# Abre o arquivo para escrita
# O modo w cria o arquivo ou apaga o conteudo anterior
with open("acessos.log", "w", encoding="utf-8") as arquivo:

    # Grava o cabecalho do sistema
    arquivo.write("SISTEMA DE AUDITORIA DE ACESSOS\n")

    # Grava o primeiro evento
    arquivo.write("INICIALIZACAO: Sistema iniciado com sucesso.\n")


# Etapa 2: Registro de Atividades (Modo Append - a)

# Abre o arquivo para adicionar novos registros
with open("acessos.log", "a", encoding="utf-8") as arquivo:

    # Adiciona cinco eventos ao arquivo
    arquivo.write("LOGIN: Usuario admin acessou o servidor.\n")
    arquivo.write("LOGIN: Usuario joao acessou o servidor.\n")
    arquivo.write("ERRO: Tentativa de acesso negada.\n")
    arquivo.write("LOGOUT: Usuario admin saiu do sistema.\n")
    arquivo.write("LOGIN: Usuario maria acessou o servidor.\n")
    arquivo.write("ERRO: Senha incorreta informada.\n")


# Etapa 3: Leitura e Inspecao Seletiva

# Abre o arquivo para leitura
with open("acessos.log", "r", encoding="utf-8") as arquivo:

    # 1 - Le a primeira linha do arquivo
    titulo = arquivo.readline()

    print("===== CABECALHO DO SISTEMA =====")
    print(titulo)

    # 2 - Le todas as linhas do arquivo
    linhas = arquivo.readlines()

    # Exibe a ultima linha registrada
    print("===== ULTIMO EVENTO REGISTRADO =====")
    print(linhas[-1])


# Etapa 4: Auditoria de Seguranca

# Define a palavra-chave que sera procurada
palavra_chave = "LOGIN"

# Inicializa o contador
contador = 0

# Abre o arquivo para leitura
with open("acessos.log", "r", encoding="utf-8") as arquivo:

    print("===== AUDITORIA DE SEGURANCA =====")

    # Percorre o arquivo linha por linha
    for linha in arquivo:

        # Verifica se a palavra-chave esta na linha
        if palavra_chave in linha:

            # Exibe o evento encontrado
            print(linha.strip())

            # Soma um ao contador
            contador += 1

# Exibe a quantidade de eventos encontrados
print("\nTotal de eventos LOGIN encontrados:", contador)
```

```python
# Etapa 1: Inicializacao do Log (Modo Escrita - w)

arquivo = None

try:
    # Tenta abrir um arquivo inexistente
    arquivo = open("servidor_inexistente.log", "r")

except FileNotFoundError:
    print("Arquivo nao encontrado. Continuando o programa.")

finally:
    # Fecha o arquivo caso tenha sido aberto
    if arquivo is not None:
        arquivo.close()


# Cria o arquivo principal
arquivo = None

try:
    arquivo = open("acessos.log", "w", encoding="utf-8")

    # Grava o cabecalho do sistema
    arquivo.write("SISTEMA DE AUDITORIA DE ACESSOS\n")

    # Grava o primeiro evento
    arquivo.write("INICIALIZACAO: Sistema iniciado com sucesso.\n")

except PermissionError:
    print("Erro: Sem permissao para escrever no arquivo.")

finally:
    # Fecha o arquivo com seguranca
    if arquivo is not None:
        arquivo.close()


# Etapa 2: Registro de Atividades (Modo Append - a)

arquivo = None

try:
    # Abre o arquivo para adicionar novos registros
    arquivo = open("acessos.log", "a", encoding="utf-8")

    # Adiciona seis eventos ao arquivo
    arquivo.write("LOGIN: Usuario admin acessou o servidor.\n")
    arquivo.write("LOGIN: Usuario joao acessou o servidor.\n")
    arquivo.write("ERRO: Tentativa de acesso negada.\n")
    arquivo.write("LOGOUT: Usuario admin saiu do sistema.\n")
    arquivo.write("LOGIN: Usuario maria acessou o servidor.\n")
    arquivo.write("ERRO: Senha incorreta - usuario bloqueado.\n")

except PermissionError:
    print("Erro: Sem permissao para adicionar registros.")

finally:
    # Fecha o arquivo com seguranca
    if arquivo is not None:
        arquivo.close()


# Etapa 3: Leitura e Inspecao Seletiva

arquivo = None

try:
    # Tenta abrir o arquivo com uma codificacao incompativel
    arquivo = open("acessos.log", "r", encoding="ascii")

    print(arquivo.read())

except UnicodeDecodeError:
    print("Erro: Codificacao do arquivo incompativel.")

except FileNotFoundError:
    print("Erro: Arquivo de log nao encontrado.")

finally:
    # Fecha o arquivo mesmo se ocorrer um erro
    if arquivo is not None:
        arquivo.close()


# Faz a leitura correta do arquivo
arquivo = None

try:
    arquivo = open("acessos.log", "r", encoding="utf-8")

    # 1 - Le somente a primeira linha
    titulo = arquivo.readline()

    print("\n===== CABECALHO DO SISTEMA =====")
    print(titulo)

    # 2 - Le as linhas restantes
    linhas = arquivo.readlines()

    # Exibe a ultima linha registrada
    print("===== ULTIMO EVENTO REGISTRADO =====")
    print(linhas[-1])

except FileNotFoundError:
    print("Erro: Arquivo de log nao encontrado.")

except UnicodeDecodeError:
    print("Erro: Nao foi possivel decodificar o arquivo.")

finally:
    # Fecha o arquivo com seguranca
    if arquivo is not None:
        arquivo.close()


# Etapa 4: Auditoria de Seguranca

arquivo = None

try:
    # Define a palavra-chave da auditoria
    palavra_chave = "LOGIN"

    # Inicializa o contador
    contador = 0

    # Abre o arquivo para leitura
    arquivo = open("acessos.log", "r", encoding="utf-8")

    print("\n===== AUDITORIA DE SEGURANCA =====")

    # Percorre o arquivo linha por linha
    for linha in arquivo:

        # Verifica se a palavra-chave esta na linha
        if palavra_chave in linha:

            # Exibe o evento encontrado
            print(linha.strip())

            # Soma um ao contador
            contador += 1

    # Exibe a quantidade de eventos encontrados
    print("\nTotal de eventos LOGIN encontrados:", contador)

except FileNotFoundError:
    print("Erro: Arquivo de log nao encontrado.")

except UnicodeDecodeError:
    print("Erro: Falha na decodificacao do arquivo.")

finally:
    # Garante o fechamento do arquivo
    if arquivo is not None:
        arquivo.close()

    print("\nAuditoria finalizada.")
```
